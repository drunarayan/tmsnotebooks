# Mathematical Progressions: Derivations, Applications, and Visualizations

This notebook provides a comprehensive guide to **Arithmetic, Geometric, and Harmonic series**. It covers algebraic derivations, finite and infinite convergence, and real-world implementations in personal finance using Python and Matplotlib.

## Part 1: Derivation of the Sum of a Geometric Series

To find the sum of the first $n$ terms of a finite geometric series ($S_n$), we write out the expansion where $a$ is the first term and $r$ is the common ratio:

$$S_n = a + ar + ar^2 + ar^3 + \dots + ar^{n-1}$$

### Step 1: Multiply by the Common Ratio
Multiply the entire equation by the common ratio $r$:

$$rS_n = ar + ar^2 + ar^3 + ar^4 + \dots + ar^n$$

### Step 2: Subtract the Equations
Subtracting the second equation from the first causes almost all intermediate terms to cancel out:

$$S_n - rS_n = (a + ar + ar^2 + \dots + ar^{n-1}) - (ar + ar^2 + ar^3 + \dots + ar^n)$$
$$S_n - rS_n = a - ar^n$$

### Step 3: Factor and Isolate $S_n$
Factor out $S_n$ on the left and $a$ on the right:

$$S_n(1 - r) = a(1 - r^n)$$

Divide by $(1 - r)$ to find the final finite geometric series sum formula:

$$S_n = \frac{a(1 - r^n)}{1 - r}$$

---

## Part 2: Extension to Infinite Geometric Series

To evaluate the sum of an infinite geometric series ($S_\infty$), we calculate the limit as the number of terms $n$ approaches infinity:

$$S_\infty = \lim_{n \to \infty} \frac{a(1 - r^n)}{1 - r}$$

If and only if the absolute value of the common ratio is less than one ($|r| < 1$), the exponential term vanishes ($ \lim_{n \to \infty} r^n = 0 $). Substituting this back into the formula yields the infinite convergence equation:

$$S_\infty = \frac{a}{1 - r}$$

### Numerical Verification Example
Let us model a geometric series with a starting term $a = 5$ and a common ratio $r = 0.5$. 

*   **Finite Sum ($n=6$):** $S_6 = \frac{5(1 - 0.5^6)}{1 - 0.5} = 9.84375$
*   **Infinite Sum Limit:** $S_\infty = \frac{5}{1 - 0.5} = 10.0$

The code cell below calculates these terms and tracks how the cumulative sum dynamically converges toward its infinite limit.

```python
import numpy as np
import matplotlib.pyplot as plt

# Define parameters
a = 5
r = 0.5
n_terms = 6

# Calculate terms and cumulative sums
n = np.arange(1, n_terms + 1)
terms = a * (r ** (n - 1))
cumulative_sums = np.cumsum(terms)

# Print specific outputs
print(f"Individual terms: {terms}")
print(f"Cumulative sum at n=6: {cumulative_sums[-1]}")
print(f"Theoretical infinite sum limit: {a / (1 - r)}")

# Plot the convergence curve
plt.figure(figsize=(9, 5))
plt.bar(n, cumulative_sums, color='#1f77b4', alpha=0.8, edgecolor='black', label='Cumulative Sum ($S_n$)')
plt.axhline(y=10.0, color='r', linestyle='--', linewidth=1.5, label='Infinite Sum Limit ($S_\infty = 10$)')

plt.title('Convergence of a Geometric Series Cumulative Sum (a=5, r=0.5)', fontsize=12, fontweight='bold', pad=15)
plt.xlabel('Number of Terms ($n$)', fontsize=10)
plt.ylabel('Cumulative Sum', fontsize=10)
plt.xticks(n)
plt.ylim(0, 12)
plt.grid(axis='y', linestyle=':', alpha=0.6)
plt.legend(loc='lower right')

for i, val in enumerate(cumulative_sums):
    plt.text(n[i], val + 0.3, f"{val:.5f}", ha='center', va='bottom', fontsize=9)

plt.tight_layout()
plt.show()
```

## Part 3: Comparing Progressions (Arithmetic, Geometric, Harmonic)

| Series Type | How Terms Progress | Mathematical Core | Infinite Convergence? |
| :--- | :--- | :--- | :--- |
| **Arithmetic** | Add a constant difference ($d$) | $S_n = \frac{n}{2}(a_1 + a_n)$ | **Never** (except when $a=0, d=0$) |
| **Geometric** | Multiply by a constant ratio ($r$) | $S_n = \frac{a(1-r^n)}{1-r}$ | **Only if** $|r| < 1$ |
| **Harmonic** | Reciprocal of counting numbers | $S_n \approx \ln(n) + \gamma$ | **Never** (diverges infinitely slowly) |

## Part 4: Real-World Personal Finance Implementations

### 1. Geometric Series: Retirement Annuity
Depositing a recurring amount ($a = \$1,000$) annually into an account earning compound interest ($8\%$) generates an annuity mapped by a geometric progression:

$$S_5 = 1000 + 1000(1.08)^1 + 1000(1.08)^2 + 1000(1.08)^3 + 1000(1.08)^4 = \$5,866.60$$

### 2. Arithmetic Series: Debt Paydown Fees
Paying an interest-free loan with a penalty transaction fee that increases by a fixed dollar amount ($d = \$2$) every month creates an arithmetic progression of fee costs:

$$S_{60} = 10 + 12 + 14 + \dots + 128 = \frac{60}{2}(10 + 128) = \$4,140$$

### 3. Harmonic Series: Dollar-Cost Averaging (DCA)
Investing a fixed fiat amount (\$100) regularly into an asset with fluctuating market values (\$10, \$20, and \$5) weights purchases toward lower prices. Your final average cost corresponds directly to the harmonic mean of the prices:

$$\text{Average Price} = \frac{3}{\frac{1}{10} + \frac{1}{20} + \frac{1}{5}} = \frac{300}{35} = \$8.57$$

The code block below generates a comparison mapping out the financial divergence between single lump-sum compounding and systemic recurring geometric additions (annuities).

```python
years = np.arange(0, 6)

# Scenario A: Single $1,000 investment compounding at 8%
single_deposit = 1000 * (1.08 ** years)

# Scenario B: Recurring $1,000 annuity investment at 8%
recurring_annuity = [0]  # End of Year 0 balance
for y in range(1, 6):
    # Balance = previous balance compounded + new end-of-year deposit
    balance = (recurring_annuity[-1] * 1.08) + 1000
    recurring_annuity.append(balance)
recurring_annuity = np.array(recurring_annuity)

# Plot the personal finance comparison
x = np.arange(len(years))
width = 0.35

plt.figure(figsize=(10, 5))
plt.bar(x - width/2, single_deposit, width, label='Single $1K Deposit', color='#ff7f0e', edgecolor='black', alpha=0.85)
plt.bar(x + width/2, recurring_annuity, width, label='Annual $1K Contributions (Annuity)', color='#2ca02c', edgecolor='black', alpha=0.85)

plt.title('Geometric Progressions in Personal Finance (8% Annual Growth)', fontsize=12, fontweight='bold', pad=15)
plt.xlabel('Timeline (Years)', fontsize=10)
plt.ylabel('Account Balance ($)', fontsize=10)
plt.xticks(x, [f'Year {y}' for y in years])
plt.grid(axis='y', linestyle=':', alpha=0.6)
plt.legend(loc='upper left')

# Add values over bars for clarity at Year 5
plt.text(5 - width/2, single_deposit[5] + 150, f"${single_deposit[5]:,.2f}", ha='center', fontsize=9, fontweight='bold')
plt.text(5 + width/2, recurring_annuity[5] + 150, f"${recurring_annuity[5]:,.2f}", ha='center', fontsize=9, fontweight='bold')

plt.ylim(0, 7000)
plt.tight_layout()
plt.show()
```
