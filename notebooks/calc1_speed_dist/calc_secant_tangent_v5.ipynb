# Calculus Lesson 2: Secant-to-Tangent Interactive Suite

This notebook introduces the formal limit definition of the derivative:

$$f'(x_0) = \lim_{\Delta x \to 0} \frac{f(x_0 + \Delta x) - f(x_0)}{\Delta x}$$

By moving $\Delta x \to 0$, you can visually watch the average rate of change (secant line) collapse into the instantaneous rate of change (tangent line) across 8 classic mathematical functions.

---

## Environment Setup

Run this setup cell first in your Jupyter Notebook to define the core plotting engine.

```python
import numpy as np
import matplotlib.pyplot as plt
import ipywidgets as widgets
from ipywidgets import interact, FloatSlider

def plot_secant_tangent(f, x_range=(-2*np.pi, 2*np.pi), padding_ratio=0.15, title="Secant to Tangent Visualization"):
    """
    Plots an interactive curve with sliders to explore secant lines transitioning to tangent lines.
    """
    x_min, x_max = x_range[0], x_range[1]
    x_dense = np.linspace(x_min, x_max, 1000)
    y_dense = f(x_dense)
    
    y_min, y_max = np.min(y_dense), np.max(y_dense)
    if y_min == y_max:
        y_min -= 1.0
        y_max += 1.0
        
    x_span = x_max - x_min
    y_span = y_max - y_min
    
    x_lims = (x_min - x_span * padding_ratio, x_max + x_span * padding_ratio)
    y_lims = (y_min - y_span * padding_ratio, y_max + y_span * padding_ratio)
    
    x0_slider = FloatSlider(
        value=(x_min + x_max) / 2.0,
        min=x_min,
        max=x_max - (x_span * 0.05),
        step=x_span / 200.0,
        description='x0:',
        continuous_update=True,
        layout=widgets.Layout(width='80%')
    )
    
    dx_slider = FloatSlider(
        value=x_span * 0.2,
        min=0.0001,
        max=x_span / 2.0,
        step=x_span / 200.0,
        description='Δx:',
        continuous_update=True,
        layout=widgets.Layout(width='80%')
    )

    def draw_frame(x0, dx):
        fig, ax = plt.subplots(figsize=(9, 5))
        
        # Plot curve
        ax.plot(x_dense, y_dense, 'b-', linewidth=2, label='f(x)')
        
        x1 = x0 + dx
        y0 = float(f(x0))
        y1 = float(f(x1))
        
        secant_slope = (y1 - y0) / dx
        
        # Points
        ax.plot([x0], [y0], 'ro', markersize=8, label='Fixed Point x0')
        ax.plot([x1], [y1], 'go', markersize=8, label='Moving Point x0 + Δx')
        
        # Extended Secant / Tangent Line
        x_ext = np.array([x_lims[0], x_lims[1]])
        y_ext = y0 + secant_slope * (x_ext - x0)
        
        if dx < (x_span * 0.01):
            ax.plot(x_ext, y_ext, 'm-', linewidth=2, label='Tangent Line')
            mode_str = f"TANGENT LIMIT (Δx -> 0)\nSlope ≈ {secant_slope:.4f}"
        else:
            ax.plot(x_ext, y_ext, 'r--', linewidth=2, label='Secant Line')
            mode_str = f"SECANT LINE\nAvg Slope = {secant_slope:.4f}"
            
        ax.set_xlim(x_lims)
        ax.set_ylim(y_lims)
        ax.set_xlabel('x')
        ax.set_ylabel('f(x)')
        ax.set_title(title)
        ax.grid(True)
        ax.legend(loc='upper right')
        
        # Display slope info
        ax.text(0.03, 0.72, f"x0 = {x0:.3f}\nΔx = {dx:.4f}\n{mode_str}", 
                transform=ax.transAxes, fontsize=10,
                bbox=dict(boxstyle='round', facecolor='white', alpha=0.85))
        
        plt.show()

    interact(draw_frame, x0=x0_slider, dx=dx_slider)
```

## 1. Parabola: $f(x) = x^2 - 4$

$$\text{Domain: } [-3, 3] \quad \implies \quad f'(x) = 2x$$

```python
# Static Preview Plot
x_p = np.linspace(-3, 3, 200)
plt.figure(figsize=(3, 1.8))
plt.plot(x_p, x_p**2 - 4, 'b-', lw=2)
plt.title("$f(x) = x^2 - 4$", fontsize=9)
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()

# Interactive Exploration
plot_secant_tangent(lambda x: x**2 - 4, x_range=(-3, 3), title="Parabola: f(x) = x² - 4")
```

## 2. Cubic Polynomial: $f(x) = x^3 - 3x$

$$\text{Domain: } [-2.2, 2.2] \quad \implies \quad f'(x) = 3x^2 - 3$$

```python
# Static Preview Plot
x_c = np.linspace(-2.2, 2.2, 200)
plt.figure(figsize=(3, 1.8))
plt.plot(x_c, x_c**3 - 3*x_c, 'b-', lw=2)
plt.title("$f(x) = x^3 - 3x$", fontsize=9)
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```

```python
# Interactive Exploration
plot_secant_tangent(lambda x: x**3 - 3*x, x_range=(-2.2, 2.2), title="Cubic: f(x) = x³ - 3x")
```

## 3. Sine Wave: $f(x) = \sin(x)$

$$\text{Domain: } [-\pi, \pi] \quad \implies \quad f'(x) = \cos(x)$$

```python
# Static Preview Plot
x_s = np.linspace(-np.pi, np.pi, 200)
plt.figure(figsize=(3, 1.8))
plt.plot(x_s, np.sin(x_s), 'b-', lw=2)
plt.title("$f(x) = \\sin(x)$", fontsize=9)
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()

# Interactive Exploration
plot_secant_tangent(np.sin, x_range=(-np.pi, np.pi), title="Sine Wave: f(x) = sin(x)")
```
