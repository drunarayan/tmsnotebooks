## Defining and Plotting Speed Functions

### Example 1: Constant Speed Profile
This example establishes baseline graph interpretation for uniform motion.

```python
# Example 1: Constant Speed
# A runner maintains a constant speed of 2.0 m/s for 10 seconds.

import matplotlib.pyplot as plt

time_pts = [0, 10]
speed_pts = [2.0, 2.0]

# Calculate total distance using rectangle area: distance = speed * time
distance = speed_pts[0] * (time_pts[1] - time_pts[0])

plt.figure(figsize=(8, 4))
plt.plot(time_pts, speed_pts, 'b-', linewidth=2, label='Speed (m/s)')
plt.fill_between(time_pts, speed_pts, color='skyblue', alpha=0.4, label=f'Distance = {distance:.1f} m')
plt.title('Exercise 1: Constant Speed Motion')
plt.xlabel('Time (s)')
plt.ylabel('Speed (m/s)')
plt.ylim(0, 3)
plt.grid(True)
plt.legend()
plt.show()
```

### Example 2: Instantaneous Step Change
This example introduces a step function where the speed drops instantly to zero, highlighting the physical impossibility of infinite acceleration.

```python
# Example 2: Instantaneous Stop (Step Change)
# A car travels at 15 m/s for 5 seconds, then instantly stops and remains stationary for 5 seconds.

import matplotlib.pyplot as plt

# Duplicate time point at t = 5 to model an instantaneous vertical drop
time_pts = [0, 5, 5, 10, 10, 20, 20]
speed_pts = [15.0, 15.0, 0.0, 0.0, 10.0, 10.0, 0]

# Distance is area under each distinct segment
# Calculate total distance using rectangle area: distance = speed * time
distance_segment1 = speed_pts[0] * (time_pts[1] - time_pts[0])
distance_segment2 = speed_pts[1] * (time_pts[2] - time_pts[1])
distance_segment3 = speed_pts[2] * (time_pts[3] - time_pts[2])
distance_segment4 = speed_pts[4] * (time_pts[4] - time_pts[3])
distance_segment5 = speed_pts[5] * (time_pts[5] - time_pts[4])


#distance_segment1 = 15.0 * 5.0
#distance_segment2 = 0.0 * 5.0
total_distance = distance_segment1 + distance_segment2 + \
                 distance_segment3 + distance_segment4 + \
                 distance_segment5

plt.figure(figsize=(8, 4))
plt.plot(time_pts, speed_pts, 'r-', linewidth=2, label='Speed (m/s)')
plt.fill_between([0, 5], [15, 15], color='orange', alpha=0.4, label=f'Distance = {total_distance:.1f} m')
plt.fill_between([10, 20], [10, 10], color='orange', alpha=0.4)
plt.title('Exercise 2: Step Change in Speed (Instantaneous Stop)')
plt.xlabel('Time (s)')
plt.ylabel('Speed (m/s)')
plt.ylim(0, 18)
plt.grid(True)
plt.legend()
plt.show()

# Conceptual Discussion Prompt for Students:
# 1. Look at the vertical line at t = 5s. What is the acceleration (slope) at this instant?
# 2. Can a real physical object change velocity from 15 m/s to 0 m/s in exactly 0.0 seconds?
```

### Example 3: Piecewise Variable Motion
This Example transitions students to piecewise continuous motion with realistic acceleration/deceleration ramps.

```python
# Example 3: Piecewise Variable Speed
# A car starts from rest, accelerates linearly to 12 m/s over 4s, cruises at 12 m/s for 4s, 
# and decelerates linearly to a stop over 2s.

import numpy as np

time_pts = [0, 4, 8, 10]
speed_pts = [0.0, 12.0, 12.0, 0.0]

total_distance = np.trapezoid(speed_pts, time_pts)

# Calculate total distance (area of trapezoid) using numpy trapz
total_distance = np.trapezoid(speed_pts, time_pts)

plt.figure(figsize=(8, 4))
plt.plot(time_pts, speed_pts, 'g-', linewidth=2, marker='o', label='Speed (m/s)')
plt.fill_between(time_pts, speed_pts, color='lightgreen', alpha=0.4, label=f'Distance = {total_distance:.1f} m')
plt.title('Exercise 3: Piecewise Motion with Continuous Speed')
plt.xlabel('Time (s)')
plt.ylabel('Speed (m/s)')
plt.ylim(0, 15)
plt.grid(True)
plt.legend()
plt.show()
```
